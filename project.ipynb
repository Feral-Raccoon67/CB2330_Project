The variable analysed in this notebook was the number of crossover events occuring on a chromosome arm during a randomly selected meiotic event. The possible values if the random variable are therefore 0, 1, 2, 3 and 4, depending on the type of crossover.

The article presents a genome-wide, high-resolution crossover map for the fruit fly *Drosophila santomea* and compares it to other closely related spscies of fruit flies. <br>

**The generative model proposed:**

A Poission distribution was proposed as a generative model for the number of crossovers occuring on one chromosome arm. This model was chosen based on the fact that the data consists of a set number of meiotic events and the number of crossovers observed for each event can be treated as a count variable. The Poisson distribution is suitable for modelling the number of discrete events that occur over a fixed unit of time, space, or distance. In this case, the fixed unit is one meiotic event, and the discrete events are crossovers.

<br>

**Parameters and assumptions:**

The parameter $θ$ (theta) was used to describe the average number of crossovers per meiotic event for a given chromosome arm. It is assumed that the crossovers occur independently at a constant average rate. Other parameters used in the model was the crossover classes, the chromosome arms and the number of observed meiotic events for each crossover class.

<br>

**The forward simulation:**

The forward simulation was performed by estimating $θ$ for each chromosome arm as the total amount of crossovers divided by the total amount of meiotic events. The Poisson model was then used for calculating the expected counts of each crossover class for each chromosome. These values were compared to the observed values from the paper.

<br>

**The parameterization/fitting:**

The parameter $θ$ was fitted to the data from the paper by using the negative log-likelihood of the Poisson distribution. The likelihood was calculated using the number of observed meiotic events in each crossover class, denoted as counts and the corresponded crossover classes, denoted as crossover_classes. Since the likelihood involves multiplying many small probabilities, causing the values to become very small, we used the negative log-likelihood. The fitted $θ$ value was obtained for each chromosome arm by finding the value that minimized the negative log-likelohood, meaning that the fitted $\theta$ represents the value that gives the best fit to the observed data according to the Poisson model.

<br>

When evaluating the results of the code written below, it appears that the estimated $θ$ values are consistent with the observed values calculated from the data in the paper. **Model limitations/how it breaks:**

A major limitation is that the model does not explain the underlying biological mechanisms for crossovers in *Drosophila santomea*. The model assumes that the crossovers occurs independently at a constant rate. However, in real meiosis, the crossover rate varies. The paper describes how biological mechanisms such as crossover interference, centromere- and telomere effects influences the distribution of crossovers. These mechanisms are not captured by the Poisson model.

<br>

Another limitation of the model is that $θ$  is calculated for the whole chromosome arm, implying that the crossover rate is homogenous across the entire chromosome arm. This means that the model breaks when the crossover rate varies across the chromosome arm. The model was also only implemented on the few chromosome arms discussed in the paper, which means it cannot be used to describe the mechanism in crossover events in other chromosome arms of the species. Lastly, the likelihood search is limited to 0.1-1.5, causing a computional limitation, since the true value could lie outside this range.

```python
# Import packages
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Describe data
crossover_classes = [0, 1, 2, 3, 4]
chromosomes = ["2L", "2R", "3L", "3R", "X"]

data = {
    "2L": [375, 336, 62, 10, 0],
    "2R": [154, 107, 20, 6, 1],
    "3L": [346, 329, 99, 7, 2],
    "3R": [294, 348, 95, 15, 4],
    "X":  [354, 331, 64, 5, 0]
}
```

# Forward running
We run the model forward by first estiamting $\theta$ for each chromosome by calculating the total amount of crossovers over the total amount of meiotic events. We then define the number of crossovers as a Poisson-distributed random variable. This allows us to calculate the expected number of meiotic events for each crossover class, which can then be compared with the observed values.

<br>

So our random variable, $X$ is the number of crossovers in a meiotic event (crossover class), which we assume follows the Poisson distribution model.

$$X \sim \text{Poisson}(θ) \tag{1} $$

The poisson probability massfunction is:  
$$P(X=x| \theta)= \frac{\theta^x e^{-\theta}}{x!} \tag{1}$$

This function gives the probability associated with each possible crossover class $X$, and  $\theta$ represents the average number of crossovers per meiotic event for a given chromosome.

<br>

The Poisson model can then be used to calculate the expected number meiotic events for each crossover class:

$$E(X) = nP(X = x | \hat{\theta}) \tag{3}$$

<br>

here $E(X)$ is the expected number of meiotic events with $X$ crossovers, $n$ is the total number of observed meiotic events, and $\hat{\theta}$ is the estimated average number of crossovers per meiotic event for the chromosome. $\hat{\theta}$ is estimated from the data by dividing the observed total amount of crossovers with the total amount of meiotic events.   


```python
def factorial(x):
    """x! as a product, 0! = 1."""
    total = 1
    for k in range(1, x + 1):
        total = total * k
    return total


def poisson(x, theta):
    """p(x | theta) = theta^x e^(-theta) / x!  the probability of x crossovers at rate theta."""
    p = (theta**x * math.exp(-theta))/ factorial(x)
    return p

def theta_estimate(crossover_classes, counts):
  """Returns the total amount of crossovers over the total amount of meiotic events"""
  total_crossovers = 0

  for i in range(len(crossover_classes)):
    x = crossover_classes[i]
    count = counts[i]
    total_crossovers += x * count

  total_events = sum(counts) #Total amount of meiotic events

  return total_crossovers/total_events
```

```python
for chromosome in chromosomes:
  counts = data[chromosome]
  theta = theta_estimate(crossover_classes, counts)
  expected = np.zeros(len(crossover_classes))

  for i in range(len(crossover_classes)):
    x = crossover_classes[i]
    expected[i] = poisson(x, theta)*sum(counts)

  print(chromosome)
  print("theta =", round(theta,2))
  print("expected =", expected.round(2))
  print("observed =", counts)

  width = 0.35

  plt.figure(figsize=(6, 4))

  plt.bar(np.array(crossover_classes) - width/2, counts,
          width=width, label="Observed", color = "hotpink")

  plt.bar(np.array(crossover_classes) + width/2, expected,
            width=width, label="Expected from Poisson", color = "pink")

  plt.xlabel("Number of crossovers")
  plt.ylabel("Number of meiotic events")
  plt.title(chromosome)
  plt.xticks(crossover_classes)
  plt.legend()

  plt.show()
```


# Backward running

The model is run backwards by using the observed crossover counts to estimate the value of $\theta$ that best describes the data. We use the Poisson model to calculate the likelihood of observing the data for different values of $\theta$.


The likelihood is calculated as

$$
L(\theta) = \prod_x P(X=x \mid \theta)^{n_x}
\tag{4}
$$

where $n_x$ is the number of observed meiotic events with $x$ crossovers.

We can then calculate the negative log-likelihood as:

$$
NLL(\theta) = -\sum_x n_x \log P(X=x \mid \theta)
\tag{5}
$$

The estimated parameter $\hat{\theta}$ is the value of $\theta$ that minimizes the negative log-likelihood:

$$
\hat{\theta} = \underset{\theta}{\operatorname{arg\,min}}\; NLL(\theta)
\tag{6}
$$

This provides an estimate of the crossover rate for each chromosome arm based on the observed data.


```python
def likelihood(crossover_classes, counts, theta):
  """L(theta) = .... p(x_i | theta)."""

  total = 1.0

  for i in range(len(crossover_classes)):

      x = crossover_classes[i]
      count = counts[i]

      total *= poisson(x, theta) ** count

  return total


def nLL(crossover_classes, counts, theta):

    total = 0.0

    for i in range(len(crossover_classes)):

        x = crossover_classes[i]
        count = counts[i]

        total -= count * math.log(poisson(x, theta))

    return total

def position_of_min(values):
  """The index of the smallest entry: the position, not the value."""
  best = 0
  for j in range(len(values)):
    if values[j] < values[best]:
      best = j

  return best
```

```python
thetas = np.linspace(0.1, 1.5, 300)
nLLs = np.zeros(len(thetas))

for i in range(len(thetas)):

    nLLs[i] = nLL(
        crossover_classes,
        data["2L"],
        thetas[i]
    )

plt.plot(thetas, nLLs, color = "hotpink")

plt.xlabel("Theta")
plt.ylabel("Negative log-likelihood")
plt.title("Negative log-likelihood for chromosome 2L")

plt.show()
```

```python
best = position_of_min(nLLs)

theta_hat = thetas[best]

print("Theta_hat =", theta_hat)
```

# Uncertainty of Theta

The uncertainty of theta is investigated by calculating the standard error. For a Poisson-distributed random variable, the standard error of $\hat{\theta}$ can be approximated as:
$$
SE(\hat{\theta}) \approx \sqrt{\frac{\hat{\theta}}{n}}
\tag{7}
$$
where $\hat{\theta}$ is the estimated average number of crossovers per meiotic event and $n$ is the total number of observed meiotic events.
The standard error describes the uncertainty associated with the estimated value of $\hat{\theta}$. A smaller standard error indicates a more precise estimate.


```python
thetas = []
ses = []

for chromosome in chromosomes:

    counts = data[chromosome]

    theta = theta_estimate(crossover_classes, counts)
    n = sum(counts)

    se = np.sqrt(theta / n)

    thetas.append(theta)
    ses.append(se)

    SE_percent = 100*se/theta

    print(chromosome)
    print("Theta =", round(theta, 3))
    print("SE =", round(se, 3))
    print("SE over estimated value =", round(SE_percent,3), "%")

plt.errorbar(chromosomes, thetas, yerr=ses, fmt="o", capsize=5, color = "hotpink")

plt.xlabel("Chromosome")
plt.ylabel("Theta_hat")
plt.title("Estimated crossover rate with uncertainty")

plt.show()
```
